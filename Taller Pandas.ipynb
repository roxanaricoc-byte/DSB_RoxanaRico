En esta actividad, se analiza un archivo con datos importantes sobre salud mental. El archivo presenta una tabla con datos de 2 mil personas con edad promedio de 38 años.

A través de pandas se explora: 
1. La actividad u ocupacion que mayor horas dedica a redes sociales y con mayor presencia de depresión en su poblacion. 
2. La relacion entre las horas dedicadas al trabajo o estudio y la interaccion social por entretenimiento. 
3. La relacion entre la edad y el tiempo diario al uso de pantallas.
4. Explorar la relacion  e impacto entre la actividad fisica y la presencia de depresion.

```python
import pandas as pd
mi_archivo = pd.read_excel("C:\\Users\\ROXANA\\Desktop\\Entregables BIT\\Todo sobre Pandas\\mental_health_ejercicio.xlsx")
total_personas = len(mi_archivo)
print("La cantidad de personas estudiadas en este archivo es de:", total_personas)
promedio_edad = mi_archivo['Age'].mean()
print("El promedio de personas estudiadas en este archivo es de:", round(promedio_edad, 0))
```

```python
#Manejo de nulos
cols_numericas = [
    'Age', 'Daily_Screen_Time', 'Social_Media_Usage', 'Night_Usage',
    'Sleep_Hours', 'Stress_Level', 'Work_Study_Hours',
    'Social_Interaction_Score', 'Caffeine_Intake'
]

for col in cols_numericas:
    mi_archivo[col] = pd.to_numeric(mi_archivo[col], errors='coerce')

# 2. Luego eliminar nulos
mi_archivo_cleaned = mi_archivo.dropna()

# 3. Convertir a int
mi_archivo_cleaned[cols_numericas] = mi_archivo_cleaned[cols_numericas].astype(int)

# 4. Ver tipos
print(mi_archivo_cleaned[cols_numericas].dtypes)

```

```python
#Agrupaciones
print("\nPorcentaje de incidencia de depresión por Ocupación:\n",)
resumen_occupation = (mi_archivo.groupby('Occupation')['Depression'].mean() * 100).round(1).astype(str) + '%' 

mi_archivo['High_Social'] = mi_archivo['Social_Media_Usage'] >  6 # umbral
resumen_high_social = (mi_archivo.groupby('Occupation')['High_Social'].mean() * 100).round(1).astype(str) + '%'
print(resumen_occupation)
print("\nPorcentaje de personas con alto uso de redes sociales (más de 3 horas diarias):\n",)
print(resumen_high_social)

##print(mi_archivo['Occupation'].value_counts())

### De acuerdo a estos datos la ocupacion con mayor incidencia de depresion es la de estudiante, con un 50% de incidencia, seguida por la ocupacion de desempleado con un 40% de incidencia.
# En cuanto al uso de redes sociales, los estudiantes tambien tienen el mayor promedio con un 60%, seguidos por los desempleados con un 50%. Esto sugiere que hay una correlacion entre el uso de redes sociales 
# y la incidencia de depresión en estas ocupaciones.
##También se observa el conteo de la muestra de cada ocupacion. 
```

```python
# Correlaciones
corr_ws_si = mi_archivo['Work_Study_Hours'].corr(mi_archivo['Social_Interaction_Score'])
corr_age_screen = mi_archivo['Age'].corr(mi_archivo['Daily_Screen_Time'])

print("\nCorrelaciones:")
print("La correlacion entre horas de trabajo/estudio y puntuacion de interaccion social es:", round(corr_ws_si, 3))
print("La correlacion entre edad y tiempo diario frente a la pantalla es:", round(corr_age_screen, 3))

###Aquí vemos que un factor no necesariamente impacta al otro, ya que la correlacion es baja.
# Esto indica que otros factores pueden estar influyendo en estas variables.
```

Ahora quise analizar si la actividad fisica presentaba algun impacto en la poblacion con depresion. Es decir, si la actividad es baja, hay mayor riesgo de depresion? 

```python
# 3.1 Porcentaje de depresión por nivel de actividad
impacto = (
    mi_archivo.groupby('Physical_Activity')['Depression']
    .mean()
    .round(3) * 100
)

correlacion = mi_archivo['Activity_Num'].corr(mi_archivo['Depression'])

# 4. Mostrar resultados
print("Porcentaje de depresión por nivel de actividad:")
print(impacto.round(1).astype(str) + '%')

print("\nCorrelación:", round(correlacion, 3))

## Con el nivel obtenido de correlacion, se puede concluir que no hay una relacion fuerte entre el nivel de actividad fisica y la incidencia de depresión en esta muestra. 
# Esto sugiere que otros factores pueden estar influyendo en la salud mental de los individuos, y que el nivel de actividad fisica por si solo no es un predictor significativo de depresión.
```

SE concluye que no hay evidencia suficiente que respalde el hecho de que dedicar mas tiempo al trabajo o al estudio resulte en mayor o menor interaccion social. Igualmente ocurre con la edad y el uso de pantallas, la actividad fisica y la depresion. 

También se observa que a pesar de esto, la informacion respalda el hecho de que los estudiantes son la poblacion más propensa al mayor uso de redes sociales y a padecer depresion.
